# Quality control and cohort selection

QC on restricted cohorts is done locally (motion scrubbing, site flags, matching). **No** participant ages, sex, or study GUIDs appear in this repository.


## Synthetic cohort table


```python
import pandas as pd
import numpy as np

rng = np.random.default_rng(42)
n = 24
participants = pd.DataFrame(
    {
        "participant_id": [f"anon_{i:03d}" for i in range(n)],
        "qc_include": rng.random(n) > 0.15,
    }
)
cohort = participants.loc[participants["qc_include"]].copy()
cohort.head()
```

For real preprocessing, keep tables local and use anonymous `participant_id` only. See [docs/data_layout.md](../docs/data_layout.md).
